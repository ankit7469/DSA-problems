```python
# Electronic Shop - print only overall max

k = list(map(int, input().split()))   # budgets
m = list(map(int, input().split()))   # keyboard prices
n = list(map(int, input().split()))   # USB prices

overall_max = -1

for u in k:
    max_sum = -1
    for i in m:
        for j in n:
            if i + j <= u:
                max_sum = max(max_sum, i + j)
    overall_max = max(overall_max, max_sum)

print(overall_max)
```
