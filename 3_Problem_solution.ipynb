```python
# Time Conversion

time = input("Enter time (hh:mm:ssAM/PM): ").strip()

period = time[-2:]        # AM or PM
t = time[:-2]             # hh:mm:ss

hour, minute, second = map(int, t.split(":"))

if period == "AM":
    if hour == 12:
        hour = 0
else:  # PM
    if hour != 12:
        hour += 12

print(f"{hour:02d}:{minute:02d}:{second:02d}")
```

```python

#  Grading Students
n=int(input())
for i in range(n):
    grade=int(input())
    next_multiple=((grade//5+1)*5)
    
    if(grade>=38):
        if((next_multiple-grade)<3):
            print(next_multiple)
        else:
            print(grade)
    else:                               # grade <38 
        print(grade)
```

```python
# Apple and Orange count

s, t = map(int, input().split())
a, b = map(int, input().split())
m, n = map(int, input().split())


count_apple=0
count_oranges=0

apples = list(map(int, input().rstrip().split()))       # Apples

oranges = list(map(int, input().rstrip().split()))      # Oranges


# for apples 

for i in apples:
    total_distance_of_apple=a+i

    if(s<=total_distance_of_apple<=t):
        count_apple +=1
print("The number of acount_apple)

# for oranges

for j in oranges:
    total_distance_of_oranges=b+j
    
    if(s<=total_distance_of_oranges<=t):
        count_oranges +=1
print(count_oranges)
```
