```python
# Cats and a mouse ---

n = int(input().strip())
for i in range(n):
    
    data = list(map(int, input().split()))
    
    x = data[0] # Cat A
    y = data[1] # Cat B
    z = data[2] # Mouse C
    
# Distance ---
    
    dist_a = abs(x - z)
    dist_b = abs(y - z)
    
#  Result 
    
    if dist_a < dist_b:
        print("Cat A")
    elif dist_b < dist_a:
        print("Cat B")
    else:
        print("Mouse C")
```
