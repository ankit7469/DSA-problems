```python
# Subarray Division (Birthday Chocolate)

n = int(input())                          # number of chocolate squares
chocolate_square = list(map(int, input().split()))
d, m = map(int, input().split())          # d = birth day (sum), m = birth month (length)

count = 0

for i in range(len(chocolate_square) - m + 1):
    if sum(chocolate_square[i:i+m]) == d:
        count += 1

print(count)
```
