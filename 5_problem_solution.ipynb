```python
# Between two sets

n, m = map(int, input().split())

a = list(map(int, input().split()))
b = list(map(int, input().split()))

count = 0

# X ko max(a) se min(b) tak check karenge
for x in range(max(a), min(b) + 1):
    
    # Condition 1: x, a ke sab elements ka multiple ho
    is_valid = True
    for i in a:
        if x % i != 0:
            is_valid = False
            break
    
    # Condition 2: x, b ke sab elements ka factor ho
    if is_valid:
        for j in b:
            if j % x != 0:
                is_valid = False
                break
    
    if is_valid:
        count += 1

print(count)
```
