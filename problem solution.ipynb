```python
# 1.  compare_triplets 

def compareTriplets(a, b):
    scoreA = 0
    scoreB = 0

    for i in range(3):
        if a[i] > b[i]:
            scoreA += 1
        elif a[i] < b[i]:
            scoreB += 1

    return [scoreA, scoreB]
```

```python
# 3. diagonal_difference

matrix = [
    [11, 2, 4],
    [4, 5, 6],
    [10, 8, -12]
]

primary = 0
secondary = 0

for i in range(3):
    primary += matrix[i][i]
    secondary += matrix[i][2 - i]

print(abs(primary - secondary))
```

```python
# 4. plus_minus

n=int(input())
import array as ar
arr=ar.array('i',[-1,-5,4,-7,0,4])

positive=0
negative=0
zero=0
for i in arr:
    if i>0:
        positive+=1
    elif i<0:
        negative+=1
    else:
        zero+=1

print(positive/n)
print(negative/n)
print(zero/n)
```
