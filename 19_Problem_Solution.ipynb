```python
# Climbing the leaderboard --

n = int(input())
ranked = list(map(int, input().split()))

m = int(input())
player = list(map(int, input().split()))

# Remove duplicates

ranked_unique = list(set(ranked))
ranked_unique.sort(reverse=True)

# find Rank

for p in player:
    rank = 1
    for score in ranked_unique:
        if p < score:
            rank += 1
        else:
            break
    print(rank)
```

```python
# 2nd method fast and reliable -- 

n = int(input())
ranked = list(map(int, input().split()))

m = int(input())
player = list(map(int, input().split()))

# Step 1: remove duplicates while keeping order
unique = []
for score in ranked:
    if not unique or unique[-1] != score:
        unique.append(score)

# Step 2: pointer from bottom of leaderboard
i = len(unique) - 1

# Step 3: process each player score
for p in player:
    while i >= 0 and p >= unique[i]:
        i -= 1
    print(i + 2)
```
