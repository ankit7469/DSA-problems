```python
# Forming a magic square

n=3
matrix=[]
for i in range(3):
    row=list(map(int,input().split()))
    matrix.append(row)

def formingMagicSquare(matrix):
    
# All 8 possible 3x3 magic squares--
    
    magic_squares = [
        [[8, 1, 6],
         [3, 5, 7],
         [4, 9, 2]],

        [[6, 1, 8],
         [7, 5, 3],
         [2, 9, 4]],

        [[4, 9, 2],
         [3, 5, 7],
         [8, 1, 6]],

        [[2, 9, 4],
         [7, 5, 3],
         [6, 1, 8]],

        [[8, 3, 4],
         [1, 5, 9],
         [6, 7, 2]],

        [[4, 3, 8],
         [9, 5, 1],
         [2, 7, 6]],

        [[6, 7, 2],
         [1, 5, 9],
         [8, 3, 4]],

        [[2, 7, 6],
         [9, 5, 1],
         [4, 3, 8]]
    ]

    min_cost = float('inf')

# Compare input matrix with each magic square --
    
    for magic in magic_squares:
        cost = 0
        for i in range(3):
            for j in range(3):
                cost += abs(matrix[i][j] - magic[i][j])
        min_cost = min(min_cost, cost)

    return min_cost

result=formingMagicSquare(matrix)
print(result)
```
