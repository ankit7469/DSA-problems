```python
# pyramid print
n=int(input())
for i in range(1,n+1):             # Because there are (n-1) spaces and (2n-1)stars in every rows
    print(" " * (n-i)+"*" *(2*i-1))
```

```python
# Inverse print 
n=int(input())
for i in range(1,n+1):
    print(" "*(n-i)+"#" *i)
```

```python
# Binary centered patterns
n = int(input())

for i in range(1, n + 1):
    print(" " * (n - i), end="")
    for j in range(1, 2*i):
        if (i + j) % 2 == 0:
            print(0, end="")
        else:
            print(1, end="")
    print()
```
