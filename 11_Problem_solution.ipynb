```python
# Bill Division (Bon Appetit)

n, k = map(int, input().split())
bill = list(map(int, input().split()))
b_charge = int(input())

total = sum(bill)
a_actual = total - bill[k]      # Anna ne ye item nahi khaya
b_actual = a_actual // 2        # Anna ka actual share

if b_charge == b_actual:
    print("Bon Appetit")
else:
    print(b_charge - b_actual)
```
